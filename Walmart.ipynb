```python
#Check the Date column
df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y")

df.head()
```

```python
daily_sales = df.groupby("Date")["Weekly_Sales"].sum()
```

```python
daily_sales
```

```python
#Plot the sales trend
plt.figure(figsize=(14,6))

plt.plot(daily_sales.index, daily_sales.values)

plt.title("Weekly Sales Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Total Weekly Sales")

plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

```python
# Calculate average sales by store
```

```python
avg_store_sales = df.groupby("Store")["Weekly_Sales"].mean().sort_values(ascending=False)

avg_store_sales
```

```python
#Top 10 stores by average weekly sales
avg_store_sales.head(10)
```

```python
#Visualize it
avg_store_sales.plot(kind="bar", figsize=(12,6))

plt.title("Average Weekly Sales by Store")
plt.xlabel("Store")
plt.ylabel("Average Weekly Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
#Cell 1 — Create a Month column
```

```python
df["Month"] = df["Date"].dt.month
```

```python
#Cell 2 — Calculate sales by month
monthly_sales = df.groupby("Month")["Weekly_Sales"].sum()

monthly_sales
```

```python
#Cell 3 — Create the visualization
monthly_sales.plot(kind="bar", figsize=(10,5))

plt.title("Total Sales by Month")
plt.xlabel("Month")
plt.ylabel("Total Weekly Sales")
plt.xticks(range(12), [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
], rotation=0)

plt.show()
```

```python
#EDA 10 — Holiday vs Non-Holiday Average Sales
```

```python
#Cell 1 — Calculate average sales
avg_holiday_sales = df.groupby("Holiday_Flag")["Weekly_Sales"].mean()

avg_holiday_sales.index = ["Non-Holiday", "Holiday"]

avg_holiday_sales
```

```python
#Cell 2 — Create the chart
avg_holiday_sales.plot(kind="bar", figsize=(8,5))

plt.title("Average Weekly Sales: Holiday vs Non-Holiday")
plt.xlabel("Week Type")
plt.ylabel("Average Weekly Sales")
plt.xticks(rotation=0)

plt.show()
```
